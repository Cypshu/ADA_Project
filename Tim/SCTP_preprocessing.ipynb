# Preprocessing - SCTP

## Imports

```python
import numpy as np

from sklearn.datasets import fetch_openml
from pandas.io.parquet import to_parquet
```

```python
data = fetch_openml(data_id=45566, as_frame=True)
data_df = data.frame
```

## Speicheroptimierung

### Float 64 --> float32

```python
float_cols = data_df.select_dtypes(include="float64").columns
float_cols = [c for c in data_df.columns if c.startswith("var")]
data_df[float_cols] = data_df[float_cols].astype("float32")

def check_downcast(df, cols, atol=1e-4):
    """
    Check if downcasting float64 to float32 is safe for the given columns in the dataframe.
    Prints the maximum and mean absolute differences, and checks if the arrays are close within the specified
    absolute tolerance (atol).
    """
    orig = df[cols].to_numpy(dtype="float64")
    down = df[cols].astype("float32").to_numpy(dtype="float64")  # zurück auf float64 für den Vergleich

    diff = np.abs(orig - down)
    print("max abs diff:  ", diff.max())
    print("mean abs diff: ", diff.mean())
    print("allclose (atol=%.0e)?" % atol, np.allclose(orig, down, atol=atol))

check_downcast(data_df, float_cols)
```
